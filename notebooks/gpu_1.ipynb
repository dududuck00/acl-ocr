```python
import json
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM, GenerationConfig, AutoModel
from tqdm import tqdm
import openai
import json
import time
import os
```

```python
os.environ["CUDA_VISIBLE_DEVICES"] = "0,1,2,3"
ocr_model_path = "../model/deepseek-ocr"
```

```python
def load_model(model_path,gpu_id):
    tokenizer = AutoTokenizer.from_pretrained(ocr_model_path, _attn_implementation='flash_attention_2', trust_remote_code=True)
    model = AutoModel.from_pretrained(
        ocr_model_path, trust_remote_code=True, use_safetensors=True
    )
    model = model.eval().cuda(f"cuda:{gpu_id}").to(torch.bfloat16)
    return tokenizer, model
```
