```python
from zai import ZhipuAiClient
import json
import base64



image_dir = "/data/home/yunhao/code/ocr/fox_data/from_text"

# 初始化客户端
client = ZhipuAiClient(api_key="a0134674aee16845354e6b804057d705.l4iiU2hVT3W8blCd")

image_url = f"{image_dir}/en_1.png"

with open(image_url, "rb") as f:
    image = base64.b64encode(f.read()).decode("utf-8")

# 调用布局解析 API
en_response = client.layout_parsing.create(
    model="glm-ocr",
    file=f"data:image/png;base64,{image}",
)

# 输出结果
print(en_response)
```
