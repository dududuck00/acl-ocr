```python
import os,torch

def occupy_memory(gpu_id=0):
    # 创建一个会占用24g显存的tensor
    free_mem = torch.cuda.mem_get_info(gpu_id)[0] - 256*1024*1024
    elem = free_mem // 4

    # torch.empty会分配显存但不初始化
    a = torch.empty((elem,), dtype=torch.float32, device=f'cuda:{gpu_id}')
    return a
```

```python
gpu_ids = [1,2,3,4,5,6,7]
tensors = []
for gpu_id in gpu_ids:
    tensors.append(occupy_memory(gpu_id=gpu_id))
```
